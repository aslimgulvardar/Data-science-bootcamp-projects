 PROJECT: Car Price Prediction System
 Advanced Regression Analysis with TensorFlow

 1. Project Objective
The goal of this project is to build an end-to-end Regression model using TensorFlow to estimate the market value of vehicles based on their technical specifications. This system simulates a real-world scenario where automotive companies automate pricing strategies using machine learning.

2. Why Regression?
Unlike classification (where we predict categories like "Yes/No"), this project deals with continuous numerical values. Since car prices can be any value within a range, we use regression to find the mathematical relationship between vehicle features and their price tags.

 3. TensorFlow Concepts Covered
Data Preparation & Normalization: Scaling features to help the neural network converge faster.
Sequential Modeling: Building a multi-layer deep learning architecture.
Regression-Specific Output: Using a linear output layer (no activation) to predict raw prices.
Loss Functions & Metrics: Utilizing Mean Squared Error (MSE) and Mean Absolute Error (MAE) to evaluate prediction accuracy.
Inference: Predicting prices for brand-new vehicle data.



```python
import numpy as np
import tensorflow as tf

n_samples = 1000
np.random.seed(42)

model_year = np.random.randint(2010, 2025, n_samples)
kilometers = np.random.randint(0, 300000, n_samples) 
engine_capacity = np.random.uniform(1.0, 5.0, n_samples)
horse_power = np.random.randint(70, 501, n_samples)
car_age = 2026 - model_year 


y = (model_year * 5000) + (horse_power * 1000) - (kilometers * 2) - (car_age * 10000)
y = y - y.min() + 50000 


X = np.column_stack((model_year, kilometers, engine_capacity, horse_power, car_age))

X_scaled = X / X.max(axis=0)
y_scaled = y / y.max()

print(" Car Data is Ready & Scaled!")
print(f"X (first car calculated): {X_scaled[0]}")
print(f"y (first car price calculated): {y_scaled[0]}")
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

print(f"training set size: {len(X_train)}")
print(f"Test Set size: {len(X_test)}")
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Input

model = Sequential()

model.add(Input(shape=(5,))) 
model.add(Dense(units=10, activation='relu'))
model.add(Dense(units=5, activation='relu'))


model.add(Dense(units=1)) 

print("model created succesfully")
```

```python
 model.compile(

    optimizer='adam',                

    loss='mean_squared_error',      

    metrics=['mae']             

)



print(" Model is ready for training") 
```

```python

history = model.fit(
    X_train, y_train,           
    epochs=100,
    batch_size=32,              
    validation_split=0.2,       
    verbose=1  )                
```

```python
# Evaluate fonksiyonu bize iki değer döndürür: 1. Loss, 2. Metrics (MAE)
test_loss, test_mae = model.evaluate(X_test, y_test)

print(f" Test MAE (approx mistake): {test_mae}")
```

```python

new_car = np.array([[2022, 15000, 1.6, 120, 4]])
new_car_scaled = new_car / X.max(axis=0)

prediction_scaled = model.predict(new_car_scaled)

final_price = prediction_scaled * y.max()

print(f"predicted car price: {final_price[0][0]:.2f} dolar")
```

Model Analysis & Insights

Most Influential Features: In this model, Model Year and Horsepower seem to have the strongest positive correlation with the price. Conversely, Kilometers and Car Age act as the primary factors for depreciation.
    
Model Strengths: The neural network is particularly good at predicting prices for "standard" vehicles that follow the general market trend (e.g., as age increases, price decreases linearly).
    
Model Limitations: The model might struggle with outliers—such as classic cars (high age but very high value) or luxury cars with small engines but high tech. Since we used a simple synthetic dataset, the model assumes a more rigid mathematical relationship.

Executive Summary

Average Prediction Error (MAE): The model predicts car prices with an average error of approximately %[0.002120441058650613]
Operational Use: This system is ready to be used as a decision-support tool for second-hand car pricing valuations.
Key Finding: Normalizing the data was crucial; without it, the high values of 'Kilometers' would have dominated the model's learning process, leading to biased results.
Future Recommendation: Adding "Brand Reputation" or "Fuel Type" as categorical features would significantly increase the model's R-squared performance.
