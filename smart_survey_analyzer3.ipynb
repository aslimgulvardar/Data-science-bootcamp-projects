```python
import random

scores = []

while len(scores) < 20:
    score = random.randint(0, 6)
    
    if score == 0:
        break
        
    if score == 6:
        continue
        
    # this line in while loop but have to outside of if
    scores.append(score)

print(scores)
```

```python
#calculate the total number of valid collected scores
total_valid_scores = len(scores)
print(f"total number of valid responses collected : {total_valid_scores}")
# WHY DID WE USE A WHILE LOOP? (Comment Line)
# Because our goal is to collect exactly 20 valid points.
# However, since we do not know when the randomly generated 6 (invalid) and 0 (exit) values will appear,
# the number of times the loop will run is uncertain.
# In such uncertain situations, a 'while' loop is more suitable than a 'for' loop.
```

```python
if len(scores) > 0:
    # this lines should be start from inside cause its starting with if
    average_scores = sum(scores) / len(scores)
    print(f"Survey average scores: {round(average_scores, 2)}")
else:
    # Bu line should be start from inside cause its starting with else
    print("The average could not be calculated because no points were collected")
```

```python
# min and max analysis
if len (scores)> 0:
    min_score= min(scores)
    max_score= max(scores)
    print(f" minumum score: {min_score}")
    print(f" maximum score: {max_score}")
else:
    print("no scores avaible for the analysis.")
```

```python
neutral = 0
unsatisfied = 0
satisfied= 0
for s in scores :
    if s >= 4:
        #if score is 4 or 5
        satisfied += 1
    elif s == 3:
        # if score equals to 3 
        neutral += 1
    else:
        #if score 1 or 2(we use else cause any other option left)
        unsatisfied = 0
        #we will print the results after loop ends                                                                                                                                                                                                                                                                                        
        
print(f"Satisfied (4-5): {satisfied}")
print(f"Neutral (3): {neutral}")
print(f"Unsatisfied (1-2): {unsatisfied}")
        
```

```python
print("--- GENERAL SUMMARY REPORT ---")
print(f"Total Valid Responses: {len(scores)}")
print(f"Average Score: {round(sum(scores)/len(scores), 2)}")
print(f"Min Score: {min(scores)}")
print(f"Max Score: {max(scores)}")
print("-" * 30)
print(f"Satisfied: {satisfied}")
print(f"Neutral: {neutral}")
print(f"Unsatisfied: {unsatisfied}")
```
