# Smart Budget Assistant day 4

**Project Goal:** The goal of this project is to develop a robust financial tracking tool that handles user inputs safely and performs analytical calculations using modular Python functions.

**Real-World Application:** This system mimics the backend logic of personal finance apps, where user data must be validated before processing to prevent system crashes.

**Python Concepts Used:**
* **Functions:** Modularizing code for reusability.
* **Error Handling (Try/Except):** Managing invalid user inputs.
* **Variable Scope:** Understanding Local vs. Global variables.
* **Flexible Arguments (*args):** Handling an unknown number of inputs.
* **Built-in Functions:** `sum`, `len`, `max`, `min`, `round`.

```python
def get_expenses():
    expense_list = []
    try:
        count = int(input("How many expenses would you like to enter? "))
        
        while len(expense_list) < count:
            try:
                value = float(input(f"Enter expense #{len(expense_list) + 1}: "))
                expense_list.append(value)
            except ValueError:
                print(" Invalid input! Please enter a numeric value for the expense.")
                
    except ValueError:
        print(" Error: Please enter a whole number for the count.")
        
    return expense_list
```

```python
def calculate_summary(expenses):
    total = sum(expenses)
    # we should check if there is any element in the list
    if len(expenses) > 0:
        avg = total / len(expenses)
    else:
        avg = 0
        
    return total, round(avg, 2)
```

```python
def flexible_total(*args):
    # here args acts like tuple 
    total = sum(args)
    count = len(args)
    print(f"Flexible Function processed {count} individual arguments.")
    return total
```

```python
def find_extremes(expenses):
    # I added control for just in case if list is empty
    if len(expenses) > 0:
        maximum = max(expenses)
        minimum = min(expenses)
        return maximum, minimum
    else:
        return 0, 0

# Tfor test(optional):
# print(find_extremes([100, 500, 250]))
```

```python
# 1. Global integer
currency = "TL"

def show_scope_demo():
    # 2. Local integer (works only in this integer)
    # In here we are trying to change currency
    currency = "USD" 
    print(f"Inside the function (Local): {currency}")

# let's call function
show_scope_demo()

# let's check output currency after function ends
print(f"Outside the function (Global): {currency}")

```

```python
def main():
    print("--- Welcome to Smart Budget Assistant v4 ---")
    
    # 1. step: put the expenses
    my_expenses = get_expenses()
    
    if not my_expenses:
        print("No expenses recorded. System exiting.")
        return

    # 2. step: summary average spent
    total_spent, average_spent = calculate_summary(my_expenses)
    
    # 3. step: max/min find
    highest, lowest = find_extremes(my_expenses)
    
    # 4. step: type results
    print("\n--- BUDGET ANALYSIS REPORT ---")
    print(f"Total Expenditure: {total_spent} {currency}")
    print(f"Average Spending  : {average_spent} {currency}")
    print(f"Highest Expense   : {highest} {currency}")
    print(f"Lowest Expense    : {lowest} {currency}")
    
    # 5. step: Flexible total  (args use for example)
    
    f_total = flexible_total(*my_expenses)
    print(f"Verified Flexible Total: {f_total}")

# PROGRAMI BAŞLAT
main()
```
