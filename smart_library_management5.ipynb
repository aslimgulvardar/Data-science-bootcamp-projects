# Smart Library Management System (OOP)

**Project Goal:** To model a real-world library system using Object-Oriented Programming (OOP) principles. 

**Why OOP?** In real life, a library isn't just a list of names. It has "Books" and "Members" that interact with each other. OOP allows us to create templates (Classes) for these entities, making the code more organized, reusable, and logical.

**Core OOP Concepts Used:**
* **Classes & Objects:** Creating blueprints for Books and Members.
* **Attributes:** Storing data inside objects (title, author, availability).
* **Methods:** Defining what an object can "do" (borrow, return).
* **Default Values:** Setting initial states (e.g., a book is available by default).
* **Inheritance:** Creating a specialized "DigitalBook" that inherits features from a standard "Book".

```python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.is_available = True 

    def get_info(self):
        status = "Available" if self.is_available else "Not Available"
        print(f"Book: {self.title} , Author: {self.author} , Status: {status}")
```

```python
class Member:
    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id

    def borrow_book(self, book):
        # First we will check if book is avaible
        if book.is_available: 
            book.is_available = False # now any other person can not borrow the book
            print(f"{self.name} borrowed the book: {book.title}")
        else:
            print(f"Sorry, {book.title} is already borrowed by someone else.")

    def return_book(self, book):
        book.is_available = True # book is returned to library
        print(f"{self.name} returned the book: {book.title}")
```

#The is_available attribute is set as a default value because every new book added to a library is assumed to be available for borrowing by default. This follows the DRY (Don't Repeat Yourself) principle, meaning we don't need to manually type True every time we register a new book.

```python
# Create a new book without specifying availability
test_book = Book("Python Basics", "data science", 2026)

# Check if it's automatically available
print(f"Book Title: {test_book.title}")
print(f"Is it available by default? -> {test_book.is_available}")
```

```python

class DigitalBook(Book):
    def __init__(self, title, author, year, file_size):
        super().__init__(title, author, year)
        self.file_size = file_size

    def download(self):
        #this method is only special for dijital book
        print(f"The digital book '{self.title}' ({self.file_size}MB) is downloading...")
```

```python

book1 = Book("Phyton Basics", "Data science", 1988)
member1 = Member("Aslim", "123")
e_book = DigitalBook("Clean Code", "Robert C. Martin", 2008, 15)


book1.get_info()
member1.borrow_book(book1)
member1.borrow_book(book1)
e_book.download()
e_book.get_info()
```

Smart Library Management System - Technical Report
1. Introduction 
The purpose of this project is to model a real-world library system using Object-Oriented Programming (OOP). Instead of using complex lists or standalone variables, we created structured "entities" like Books and Members. This approach makes the code more scalable and easier to maintain.

2. Core OOP Concepts Used 
Class & Objects: We defined Book and Member as blueprints (Classes). The actual books we created (like The Alchemist) are the "Instances" (Objects) of these classes.

Encapsulation (Attributes): Every book carries its own data (title, author, year) within itself. This ensures that each object manages its own information.

Default Values: By setting is_available = True in the __init__ method, we automated the availability status for every new book, following the DRY (Don't Repeat Yourself) principle.

Inheritance: We created a DigitalBook class that inherits from the Book class. This allowed us to reuse existing code while adding specific features like file_size and a download() method.

3. Interaction Logic 
The heart of the project is the interaction between the Member and Book objects.

The borrow_book method performs a logical check: It only allows a transaction if the book's is_available attribute is True.

Once borrowed, the system updates the book's state to False, preventing others from taking the same copy. This mimics a real-life library checkout process.

4. Conclusion 
This project demonstrates how OOP can transform real-world scenarios into functional code. By using inheritance and method interactions, we built a system that is organized and logical.
