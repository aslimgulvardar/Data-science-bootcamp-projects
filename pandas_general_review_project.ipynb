# Project: Smart Customer Data Analysis System

 Purpose of the Project
The goal of this project is to analyze customer spending behavior and profile data using Pandas. 
By transforming raw data into structured Series and DataFrames, we aim to:
 Track monthly expenses efficiently.
 Manage customer information (City, Membership Type) in a tabular format.
 Use advanced indexing (Multi-index) to categorize data for better decision-making.

 Why Pandas ?
 Labels: Unlike NumPy, Pandas allows us to use names (indices) instead of just numbers.
Flexibility: It handles different data types (strings, integers, floats) together.
Easy Analysis: It provides built-in methods for quick statistical summaries.

 Covered Topics
 Pandas Series & DataFrames
 Index Resetting & Setting
 Multi-Indexing (Hierarchical Data)

```python
import pandas as pd


spending_values = [3200, 4500, 2800] 


months = ["September", "November", "December"]


monthly_expenses = pd.Series(data=spending_values, index=months)

print(" Monthly Spending Series ")
print(monthly_expenses)
```

```python

print(f"Spending Values: {monthly_expenses.values}")

print(f"Months (Index): {monthly_expenses.index}")

print(f"Data Type: {monthly_expenses.dtype}")

print(f"Average Monthly Spending: {monthly_expenses.mean()}")
```

```python

data = {
    "Customer_name": ["Aslim", "Duru", "Oyku"], 
    "City": ["London", "New York", "Istanbul"],
    "Monthly_spending": [3900, 2800, 4500],
    "Membership_type": ["Silver", "Gold", "Diamond"] 
}


df = pd.DataFrame(data)


print("--- Customer Data Table ---")
print(df)
```

```python

df_new = df.set_index("Customer_name")

aslim_data = df_new.loc["Aslim"]

print("Data for Aslim ")
print(aslim_data)
```

```python

df_reset = df_new.reset_index()

df_city = df_reset.set_index("City")

print(" Table Indexed by City ")
print(df_city)

london_customers = df_city.loc["London"]
print(" Customers in London ")
print(london_customers)
```

```python

df_multi = df_reset.set_index(["City", "Membership_type"])

print("Multi-Indexed Table ")
print(df_multi)
```

```python

print(f"Total Customers: {len(df)}")
print(f"Highest Monthly Spending: {df['Monthly_spending'].max()}")
print(f"City with Most Customers: {df['City'].value_counts().idxmax()}")
```

 Technical Report: Smart Customer Data Analysis Pandas
1. Data Structure: Series vs. DataFrame
In this project, we transitioned from basic NumPy arrays to Pandas' more sophisticated structures.

Series: Used for time-series data (Monthly Expenses). It allowed us to map values to specific months using custom indices, making the data human-readable.

DataFrame: Used for our primary customer database. This allowed us to store diverse data types (strings for names/cities, integers for spending) in a single, unified table.

2. The Strategic Importance of Indexing
We moved beyond default integer indexing (0, 1, 2...) to demonstrate how indexing acts as a unique identifier.

By setting Customer_name as the index, we optimized the "Search" process using .loc.

This transformation turns a simple list into a searchable database where individual profiles can be extracted instantly.

3. Hierarchical Data (Multi-Indexing)
The implementation of a Multi-Index (City + Membership_type) solved the problem of data categorization.

The Problem: In large datasets, finding a "Gold" member in "London" requires complex filtering.

The Solution: Multi-indexing creates a hierarchy, allowing the system to "drill down" through layers of data (e.g., City -> Membership -> Customer) without losing performance.

4. Data Aggregation & Insights
Using Pandas' built-in analytical methods (.max(), .mean(), .idxmax()), we successfully performed high-level business intelligence:

Identified the Highest Spender to target for loyalty programs.

Determined the Primary Market (City) with the highest customer density, providing a data-driven basis for marketing focus.

5. Conclusion
Pandas has proven to be the superior tool for structural data analysis. While NumPy handles the heavy mathematical lifting, Pandas provides the organizational framework necessary for real-world business applications and complex data storytelling.
