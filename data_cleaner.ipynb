```python
height_cm = float(parts[2]) * 100
print(height_cm)
```

Smart Data Cleaner Technical Report
### 1. Introduction

The main objective of this project is to standardize unstructured and “dirty” data obtained from external sources (such as forms, surveys, or raw text files) by utilizing Python’s text processing capabilities. In data science, the principle of *“Garbage In, Garbage Out”* applies. Therefore, cleaning the data before moving on to the analysis stage is crucial for the accuracy and sustainability of the system.

---

### 2. Python Concepts Used

While implementing the project, the following concepts were used within a logical framework:

* **Variables:** After splitting the raw data into parts, variables were used to store each piece of information (name, age, height) in memory and to perform independent operations on them.

* **String Data Type and Methods:** Used to manipulate text-based data.

  * `strip()` removed surrounding noise (whitespace),
  * `split()` transformed complex structures into meaningful components,
  * `title()` and `lower()` ensured visual consistency.

* **String Slicing:** Used to focus on a specific part of a data group (e.g., extracting the user code from an email address) and retrieve it precisely.

* **Type Conversions (int/float):** Numeric values initially interpreted as text were converted into numerical formats to enable mathematical operations (such as age calculations or unit conversions for height).

---

### 3. Data Cleaning Process

**Problems in Raw Data:**
The raw data contained inconsistent use of uppercase and lowercase letters, unnecessary spaces, and numerical values embedded within text. In this form, the data could neither be stored in a database nor used for mathematical analysis.

**Applied Solutions:**

* **Parsing:** The data was split using semicolons (`;`) as separators.
* **Normalization:** Irregular letter patterns in names were converted into a standardized format (Title Case).
* **Error Handling:** Whitespace around numerical data was removed to prevent type errors (*ValueError*).
* **Data Derivation:** A new “user code” was generated from raw email data using slicing techniques.

**Impact of Dirty Data:**
If uncleaned data is entered into the system, it may result in recording the same person under different names (e.g., *Ahmet* vs *AHMeT*), cause the program to crash during age calculations, or produce incorrect results in height/weight analyses.

---
