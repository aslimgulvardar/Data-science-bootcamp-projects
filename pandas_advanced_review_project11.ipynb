 Project: Smart Sales & Customer Analytics System

 Project Overview
The primary objective of this project is to build an end-to-end sales and customer analysis system using Pandas. In a real-world business scenario, data is often scattered across multiple sources and contains inconsistencies. This project simulates that environment by integrating disparate datasets (Customer profiles, Sales records, and Regional reports) into a unified, clean, and actionable business intelligence report.

Why Pandas?

Data Cleaning:Handling missing values (NaN) which are inevitable in raw business data.
Analytical Power: Using complex aggregations and transformations to derive insights.
Exportability: Bridging the gap between Python and business users by generating Excel outputs.

 Covered Pandas Topics
Missing Data Management: Detection and imputation strategies using isna, fillna, and dropna.
Data Merging: Implementing relational logic with Merge (Inner/Left) and Concat.
Hierarchical Analysis: Multi-dimensional grouping with groupby and agg.
Advanced Operations: Applying logic to datasets via apply(), lambda functions, and conditional filtering.
Excel Integration: Reading from and writing to professional report formats (read_excel, to_excel,).

```python
import pandas as pd
import numpy as np
```

```python
import pandas as pd
import numpy as np

customers_data = {
    "CustomerID": [101, 102, 103, 104, 105],
    "CustomerName": ["Samantha", "Carrie", "Charlotte", "Miranda", "Stanford"],
    "City": ["New York", "Paris", "Istanbul", "London", "Tokyo"],
    "Age": [45, 38, np.nan, 36, 40] 
}
customers = pd.DataFrame(customers_data)

sales_data = {
    "SaleID": [1, 2, 3, 4, 5],
    "CustomerID": [101, 102, 103, 104, 106], 
    "Product": ["Shoes", "Dress", "Bag", "Coat", "Shoes"],
    "Amount": [1000, 2300, np.nan, 900, 1500] 
}
sales = pd.DataFrame(sales_data)


regions_data = {
    "City": ["New York", "Paris", "London", "Tokyo"],
    "Region": ["North America", "Europe", "Europe", "Asia"]
}
regions = pd.DataFrame(regions_data)

print(" Customers Table ")
print(customers)
print(" Sales Table ")
print(sales)
```

```python

customers["Age"] = customers["Age"].fillna(customers["Age"].mean())
sales = sales.dropna(subset=["Amount"])

print(" Cleaned Customers ")
print(customers)
print(" Cleaned Sales ")
print(sales)
```

```python

df_combined = pd.merge(customers, sales, on="CustomerID", how="inner")
df_final = pd.merge(df_combined, regions, on="City", how="left")

print("Final Merged Table")
print(df_final)
```

```python

city_sales = df_final.groupby("City")["Amount"].agg(["sum"])
region_analysis = df_final.groupby("Region")["Amount"].agg(["mean"])
product_stats = df_final.groupby("Product")["Amount"].agg(["count"])

print(" City Based Sales ")
print(city_sales)
print(" Region Based Analysis ")
print(region_analysis)
```

```python

new_sales_data = {
    "SaleID": [6, 7],
    "CustomerID": [101, 105],
    "Product": ["Watch", "Belt"],
    "Amount": [1200, 450]
}
sales_feb = pd.DataFrame(new_sales_data)
all_sales_combined = pd.concat([sales, sales_feb], axis=0)


all_sales_combined = all_sales_combined.reset_index(drop=True)

print("All Sales Combined (Concat) ")
print(all_sales_combined)
```

```python

high_sales = df_final[df_final["Amount"] > 1000]
df_final["Total_Price"] = df_final["Amount"] * 1.20
df_final["CustomerName"] = df_final["CustomerName"].apply(lambda x: x.upper())

print(" High Sales (Over 1000) ")
print(high_sales)
print("Final Table with VAT and UpperCase Names ")
print(df_final)
```

```python
df_final.to_excel("final_sales_report.xlsx", index=False)
print("Report is finished please check your files. 📁")
```
