 3. Advanced Matplotlib Tools Used

Visual Aesthetics:** Customizing backgrounds, grid styles, and color consistency for a premium look.

```python
import numpy as np
import pandas as pd

months = np.array(['May', 'June', 'July', 'August'])
monthly_sales = np.random.randint(2000, 8000, size=4)

product_data = {
    "Products": np.array(["Bag", "Shoes", "Scarf", "Watch"]),
    "Revenue": np.array([4500, 6200, 3100, 5800])
}


regions = np.array(['Europe', 'Asia', 'Africa', 'America'])
region_performance = np.array([85, 92, 70, 88]) 


customer_ages = np.random.normal(35, 10, 50)
spending_score = customer_ages * 1.5 + np.random.randint(10, 50, 50) 

print(" Advanced Data Preparation Complete")
print(f"Target Months: {months}")
print(f"Top Product: {product_data['Products'][np.argmax(product_data['Revenue'])]}")
```

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(2, 2, figsize=(16, 12))


fig.suptitle('EXECUTIVE STRATEGY DASHBOARD - 2026 Q3', fontsize=22, fontweight='bold', color='#2C3E50')

ax[0, 0].plot(months, monthly_sales, color="hotpink", marker="o", linestyle="-.", linewidth=3)
ax[0, 0].set_title("Monthly Sales Growth", fontsize=14, fontweight='bold')

ax[0, 0].annotate('Peak Season', xy=('August', monthly_sales[3]), xytext=('July', monthly_sales[3]+500),
             arrowprops=dict(facecolor='black', shrink=0.05), fontsize=4)
ax[0, 0].grid(True, linestyle='--', alpha=0.4) 


bars = ax[0, 1].bar(product_data["Products"], product_data["Revenue"], color="#3498DB")
ax[0, 1].set_title("Revenue by Product Category", fontsize=4, fontweight='bold')
ax[0, 1].bar_label(bars, padding=3)
ax[1, 0].scatter(customer_ages, spending_score, color="purple", alpha=0.6, edgecolors='w', s=100)
ax[1, 0].set_title("Customer Age vs. Spending Score", fontsize=4, fontweight='bold')
ax[1, 0].set_xlabel("Customer Age")
ax[1, 0].set_ylabel("Spending Score")


ax[1, 1].pie(product_data["Revenue"], labels=product_data["Products"], 
        autopct='1.1f', colors=["#FF9999", "#66B3FF", "#99FF99", "#FFCC99"], 
        startangle=140, explode=(0.1, 0, 0, 0)) 
ax[1, 1].set_title("Revenue Share %", fontsize=4, fontweight='bold')


plt.tight_layout(rect=[0, 0.03, 1, 0.95]) 
plt.show()
```
