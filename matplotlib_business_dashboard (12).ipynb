

1. Project Objective
The primary goal of this project is to transform raw sales data into a high-level Visual Dashboard. By utilizing Matplotlib, we translate complex datasets into intuitive charts that allow management to monitor business health at a glance and make data-driven decisions.

2. Why Data Visualization?
Rapid Analysis: Identifies trends and patterns much faster than scanning raw tables.
Error Detection: Visual outliers immediately highlight potential data issues or performance drops.
Effective Communication: Simplifies technical data for non-technical stakeholders and executives.

3. Matplotlib Core Competencies Covered
Matplotlib Foundations: Figure and Axes management.
Comparative Analysis: Utilizing Bar Charts to contrast product performance.
Trend Analysis:Utilizing Line Plots to track monthly time-series data.
Visualization Customization: Enhancing readability with titles, labels, colors, markers, and legends.
Dashboard Architecture: Implementing Subplots (2x2) to consolidate multiple insights into a single view.



```python
import numpy as np
import pandas as pd

months = np.array(['September', 'October', 'November', 'December'])

monthly_sales = np.random.randint(2000, 6001, size=4)

product_data = {
    "Products": np.array(["Laptop", "Phone", "Headphones"]),
    "Revenue": np.array([1000, 2000, 4000])
}

regions = np.array(['North', 'South', 'East', 'West'])
region_sales = np.random.randint(5000, 15000, size=4)

print("Data Preparation Complete!")
print(f"Months: {months}")
print(f"Sales: {monthly_sales}")
```

```python

plt.figure(figsize=(10, 5)) 
plt.plot(months, monthly_sales, marker='o', linestyle='--', color='purple', label='Monthly Sales')

plt.title('Monthly Sales Trend ', fontsize=14)
plt.xlabel('Months')
plt.ylabel('Sales Amount ')

plt.show()
```

```python

plt.figure(figsize=(10, 5))
plt.bar(product_data["Products"], product_data["Revenue"], color='pink', label='Revenue per Product')

plt.title("Product Sales & Revenue Analysis", fontsize=10)
plt.xlabel("Products")
plt.ylabel("Revenue ")


plt.show()
```

```python

plt.figure(figsize=(10,5))
plt.plot(months, monthly_sales, 
         color="hotpink", 
         marker="*", 
         markersize=10, 
         linestyle="--", 
         linewidth=2, 
         label="Monthly Sales")

plt.title("Business Performance", fontsize=14, color="deeppink", style="italic", fontweight="bold")
plt.xlabel("Months of Business")
plt.ylabel("Revenue ", fontsize=11)

plt.grid(True, linestyle=":", alpha=0.5)
plt.legend(loc="upper right") 

plt.show()
```

```python

fig, ax = plt.subplots(2, 2, figsize=(15, 10))

ax[0, 0].plot(months, monthly_sales, color="hotpink", marker="o", linestyle="--")
ax[0, 0].set_title("Monthly Sales Trend")
ax[0, 0].set_xlabel("Months")
ax[0, 0].set_ylabel("Sales ($)")
ax[0, 0].grid(True, alpha=0.3)


ax[0, 1].bar(product_data["Products"], product_data["Revenue"], color="skyblue")
ax[0, 1].set_title("Product Performance")
ax[0, 1].set_xlabel("Products")
ax[0, 1].set_ylabel("Revenue ")


ax[1, 0].bar(regions, region_sales, color="salmon")
ax[1, 0].set_title("Regional Distribution")
ax[1, 0].set_xlabel("Regions")
ax[1, 0].set_ylabel("Sales ($)")


ax[1, 1].pie(product_data["Revenue"], labels=product_data["Products"], autopct='1.1f', colors=["pink", "lightblue", "lightgreen"])
ax[1, 1].set_title("Revenue Share by Product")

plt.tight_layout()
plt.show()
```
